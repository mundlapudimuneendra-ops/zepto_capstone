CLEAN PRICE

```python
df["price"] = (
    df["price"].str.replace("Â£","",regex=False)
    .astype(float)
)
```

```python
df.rename(columns = {"price" : "price_gbp"}, inplace = True)
```

CONVERT RATING

```python
rating = {
    "One" : 1,
    "Two" : 2,
    "Three" : 3,
    "Four" : 4,
    "Five" : 5
}

df["star_rating"] = df["star_rating"].map(rating)
```

```python
df["star_rating"].head()
```

ADD INR PRICE

```python
GBP_TO_INR = 105
df["price_inr"] = df["price_gbp"] * GBP_TO_INR
```

CONVERTS BOOLEAN

```python
df["availability"].unique()
```

```python
df["availability"] = (
    df["availability"]
    .str.strip()
    .str.contains("in stock", case = False)
)
```
